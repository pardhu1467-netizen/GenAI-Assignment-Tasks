Task 1: Conceptual Understanding

1. Difference between "Love" and "love" in NLP
- NLP models treat words as tokens.
- "Love" and "love" are different tokens due to case sensitivity.
- This leads to data sparsity (same meaning, different representation).
- Hence, we convert text to lowercase to ensure consistency.
  
2. What happens if stopwords are not removed?
- Increased noise in data
- Higher dimensionality (more unnecessary tokens)
- Reduced model efficiency
- Important signals may get diluted
  
3. When removing stopwords can be harmful?
- Sentiment Analysis
"I am not happy" → removing "not" changes meaning completely
- Question Answering / Chatbots
"What is the time?" → removing "what" breaks intent

4. Stemming vs Lemmatization
| Feature	| Stemming	|Lemmatization |
|-----------|-----------|--------------|
|Approach	|Rule-based	|Dictionary-based|
|Output|	Root (may not be valid word)|	Valid word|
|Example	|"happiness" → "happi"	|"happiness" → "happy"|
|Accuracy	|Lower|	Higher|

```python
import re

def preprocess_text(text):
    if not isinstance(text, str) or text.strip() == "":
        return [], ""

    # Remove URLs and emails
    text = re.sub(r"http\S+|www\S+|https\S+", "", text)
    text = re.sub(r"\S+@\S+", "", text)

    # Remove numbers
    text = re.sub(r"\d+", "", text)

    # Lowercase
    text = text.lower()

    # Handle repeated characters (soooo → so)
    text = re.sub(r'(.)\1{2,}', r'\1', text)

    # Remove special characters (keep alphabets & spaces)
    text = re.sub(r"[^a-z\s]", "", text)

    # Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    # Tokenization
    tokens = text.split()

    # Remove short tokens (≤2) except "no", "not"
    tokens = [word for word in tokens if len(word) > 2 or word in ["no", "not"]]

    clean_sentence = " ".join(tokens)

    return tokens, clean_sentence
```

```python
test_sentences = [
    "Get 100% FREE access now!!!",
    "I absolutely looooved this product 😍😍",
    "Worst service ever... 0/10",
    "Call me at 9876543210",
    "This is THE best course!!!",
    "Visit https://openai.com now!",
    "Nooooo this is baaad!!!",
    "OK OK OK I got it",
    "Win $$$ now!!! Limited offer!!!",
    "I am not happy with this"
]

results = []

for sentence in test_sentences:
    tokens, clean_sentence = preprocess_text(sentence)
    print("Original:", sentence)
    print("Tokens:", tokens)
    print("Cleaned:", clean_sentence)
    print("-" * 50)
    
    results.append((sentence, tokens, clean_sentence))
```

```python
import numpy as np

analytics = []

for sentence, tokens, _ in results:
    total_tokens = len(tokens)
    unique_tokens = len(set(tokens))
    avg_length = np.mean([len(t) for t in tokens]) if tokens else 0

    analytics.append({
        "sentence": sentence,
        "total_tokens": total_tokens,
        "unique_tokens": unique_tokens,
        "avg_token_length": avg_length
    })

for item in analytics:
    print(item)
```

```python
from collections import Counter

all_tokens = []

for _, tokens, _ in results:
    all_tokens.extend(tokens)

counter = Counter(all_tokens)

print("Top 10 frequent words:")
print(counter.most_common(10))

print("\nTop 5 least frequent words:")
print(counter.most_common()[-5:])
```

```python
def full_pipeline(text_list):
    all_tokens = []
    clean_sentences = []

    for text in text_list:
        tokens, clean_sentence = preprocess_text(text)
        all_tokens.extend(tokens)
        clean_sentences.append(clean_sentence)

    return {
        "tokens": all_tokens,
        "clean_sentences": clean_sentences
    }
```
